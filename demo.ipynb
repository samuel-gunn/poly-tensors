```python
import torch
from PolyTensor import PolyTensor

X = PolyTensor([0., 1.], degree=5)
w_init = torch.tensor([1., -1., 2.])
w = PolyTensor(w_init + X, requires_grad=True)

opt = torch.optim.SGD([w], lr=0.1, foreach=False)

for _ in range(3):
    opt.zero_grad()
    loss = (w ** 3).sum()
    loss.backward()
    opt.step()
```
